```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the data from Data.csv
# Note: Ensure Data.csv is in the same folder as your script
df = pd.read_csv('Data.csv')

# 2. Sum 'order_cnt' for each branch (grouped by ISIN and Branch Name)
# This handles cases where different ISINs might share branch names
branch_totals = df.groupby(['isin', 'branch_name'])['order_cnt'].sum().reset_index()

# 3. Sort by total count for a clearer visualization
branch_totals = branch_totals.sort_values(by='order_cnt', ascending=True)

# 4. Visualization: Horizontal Bar Chart
plt.figure(figsize=(10, 8))
plt.barh(branch_totals['branch_name'], branch_totals['order_cnt'], color='skyblue')
plt.xlabel('Total Order Count')
plt.ylabel('Branch Name')
plt.title('Total Order Count per Branch')
plt.tight_layout()

# Save the visualization
plt.savefig('branch_visualization.png')

# 5. Save the summary to a new CSV file
branch_totals.to_csv('branch_summary.csv', index=False)

# Display the summary table
print(branch_totals)
```
