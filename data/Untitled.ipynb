```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the data from Data.csv
# Note: Ensure Data.csv is in the same folder as your script
df = pd.read_csv('Data.csv')

# 2. Sum 'order_cnt' for each branch (grouped by ISIN and Branch Name)
# This handles cases where different ISINs might share branch names
branch_totals = df.groupby(['broker_code','broker_name', 'branch_name'])['order_cnt'].sum().reset_index()
branch_totals2 = df.groupby(['broker_name'])['order_cnt'].sum().reset_index()
```

```python
# 3. Sort by total count for a clearer visualization
branch_totals = branch_totals.sort_values(by='order_cnt', ascending=False)
branch_totals2 = branch_totals2.sort_values(by='order_cnt', ascending=False)
```
